```python
import pandas as pd

tickets = {
    "Movie": [
        "Avatar", "Titanic", "Inception", "Avatar",
        "Frozen", "Titanic", "Avatar", "Frozen",
        "Inception", "Titanic", "Frozen", "Avatar"
    ],
    "City": [
        "Baku", "Ganja", "Baku", "Sumgait",
        "Ganja", "Baku", "Ganja", "Sumgait",
        "Sumgait", "Ganja", "Baku", "Baku"
    ],
    "Tickets Sold": [
        120, 90, 75, 100, 110, 130,
        140, 85, 95, 105, 125, 160
    ]
}

df_tickets = pd.DataFrame(tickets)

print("All ticket sales:")
print(df_tickets)

print("\nTickets sold per movie:")
print(df_tickets.groupby("Movie")["Tickets Sold"].sum())

print("\nTickets sold per city:")
print(df_tickets.groupby("City")["Tickets Sold"].sum())

movie_sales = df_tickets.groupby("Movie")["Tickets Sold"].sum()

print("\nMovie with the highest sales:")
print(movie_sales.idxmax())
print("Total tickets sold:", movie_sales.max())
```
