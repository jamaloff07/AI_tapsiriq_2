```python
import pandas as pd

# Create a DataFrame
books = {
    "Title": [
        "Python Basics",
        "Data Science",
        "AI for Beginners",
        "Machine Learning",
        "Web Development",
        "Deep Learning"
    ],
    "Author": [
        "John Smith",
        "Alice Brown",
        "David Wilson",
        "Emma Davis",
        "Michael Lee",
        "Sarah Johnson"
    ],
    "Year Published": [2014, 2018, 2020, 2016, 2012, 2023],
    "Price": [25.5, 40.0, 30.0, 45.5, 20.0, 55.0]
}

df_books = pd.DataFrame(books)

print("All books:")
print(df_books)

print("\nAverage price:")
print(df_books["Price"].mean())

print("\nBooks published after 2015:")
print(df_books[df_books["Year Published"] > 2015])

print("\nBooks sorted by price:")
print(df_books.sort_values(by="Price", ascending=True))
```
