```python
import pandas as pd

employees = {
    "Name": [
        "Ali", "Aysel", "Kamran", "Ali",
        "Nigar", "Aysel", "Kamran", "Nigar"
    ],
    "Project": [
        "Website", "AI System", "Database", "AI System",
        "Website", "Database", "Website", "AI System"
    ],
    "Hours": [20, 35, 25, 30, 15, 20, 40, 25]
}

df_employees = pd.DataFrame(employees)

print("Employee project data:")
print(df_employees)

print("\nTotal hours for each employee:")
print(df_employees.groupby("Name")["Hours"].sum())

print("\nTotal hours for each project:")
print(df_employees.groupby("Project")["Hours"].sum())

employee_hours = df_employees.groupby("Name")["Hours"].sum()

print("\nEmployee with the most hours:")
print(employee_hours.idxmax())
print("Total hours:", employee_hours.max())
```
