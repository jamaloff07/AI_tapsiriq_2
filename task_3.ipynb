```python
import pandas as pd

foods = {
    "Product": [
        "Chicken", "Rice", "Salmon", "Bread", "Almonds",
        "Eggs", "Cheese", "Potato", "Beef", "Oats"
    ],
    "Category": [
        "Meat", "Grains", "Fish", "Grains", "Nuts",
        "Dairy", "Dairy", "Vegetables", "Meat", "Grains"
    ],
    "Calories": [330, 280, 350, 265, 579, 155, 402, 180, 380, 389],
    "Protein": [31, 6, 25, 9, 21, 13, 25, 4, 26, 17]
}

df_foods = pd.DataFrame(foods)

print("All food products:")
print(df_foods)

print("\nFoods with calories above 300:")
print(df_foods[df_foods["Calories"] > 300])

print("\nAverage protein by category:")
print(df_foods.groupby("Category")["Protein"].mean())

print("\nFoods sorted by calories:")
print(df_foods.sort_values(by="Calories", ascending=False))
```
