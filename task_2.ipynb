```python
import pandas as pd

df_orders = pd.read_csv("orders.csv")

print("First 10 rows:")
print(df_orders.head(10))

print("\nOrders per customer:")
print(df_orders["Customer"].value_counts())

print("\nMaximum order amount:")
print(df_orders["Amount"].max())

print("\nMinimum order amount:")
print(df_orders["Amount"].min())

print("\nTotal amount of all orders:")
print(df_orders["Amount"].sum())
```
