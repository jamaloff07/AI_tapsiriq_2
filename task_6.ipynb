```python
import pandas as pd

customers = {
    "ID": [1, 2, 3, 4, 5],
    "Name": ["Farid", "Aysel", "Kamran", "Maryam", "Murad"],
    "City": ["Baku", "Ganja", "Sumgait", "Baku", "Shaki"]
}

df_customers = pd.DataFrame(customers)

orders = {
    "Customer ID": [1, 2, 1, 3, 4, 2, 5, 3],
    "Order": [
        "Laptop", "Phone", "Mouse", "Keyboard",
        "Monitor", "Headphones", "Tablet", "Mouse"
    ],
    "Cost": [1200, 800, 25, 70, 300, 100, 450, 30]
}

df_customer_orders = pd.DataFrame(orders)

print("Customers:")
print(df_customers)

print("\nOrders:")
print(df_customer_orders)

df_merged = pd.merge(
    df_customers,
    df_customer_orders,
    left_on="ID",
    right_on="Customer ID"
)

print("\nMerged DataFrame:")
print(df_merged)

df_total_cost = (
    df_merged.groupby(["ID", "Name", "City"], as_index=False)["Cost"]
    .sum()
)

df_total_cost = df_total_cost.rename(
    columns={"Cost": "Total Cost"}
)

print("\nTotal cost for each customer:")
print(df_total_cost)
```
